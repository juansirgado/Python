```python
#-----------------------------------------------------------
import os as os
import pytz as dtz
import pandas as pd
from datetime import datetime
from influxdb import InfluxDBClient
#-----------------------------------------------------------
```

```python
#-----------------------------------------------------------
dbHost = "172.16.81.91"
dbPort = "8086"
dbName = "mainframe"
dbTable = "mycpuload"
myDtTm = datetime.now(tz=dtz.UTC).strftime('%Y-%m-%dT%H:%M:%SZ')
print(myDtTm)
#-----------------------------------------------------------
```

```python
#-----------------------------------------------------------
infClient = InfluxDBClient(host=dbHost, port=dbPort)
infClient.switch_database(dbName)
file_path = os.getcwd() + "\InfluxDB.csv"
print(file_path)
#-----------------------------------------------------------
```

```python
#-----------------------------------------------------------
csvReader = pd.read_csv(file_path)
myCol = csvReader.keys()
print(csvReader.shape)
csvReader.head()
#-----------------------------------------------------------
```

```python

#-----------------------------------------------------------
# Monta e grava a última linha do arquivo CSV no InfluxDB
# Para gravar tudo, colocar o write_points dentro do loop
#-----------------------------------------------------------
for row_index, myRow in csvReader.iterrows() : 
    myJson = [{
       "time": myDtTm,
       "measurement": dbTable,
       "tags": {
          myCol[0]: myRow[0],
          myCol[1]: myRow[1]
          },
       "fields": {
          myCol[2]: myRow[2],
          myCol[3]: myRow[3],
          myCol[4]: myRow[4],
          myCol[5]: myRow[5]
          }
    }]
#-----------------------------------------------------------
print(myJson)
#-----------------------------------------------------------
```

```python
#-----------------------------------------------------------
infClient.write_points(myJson)
#-----------------------------------------------------------
```
