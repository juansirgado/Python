```python

for col_desc in curs.description: 
    print (col_desc[0], end="; ")
print("")
```

```python
row_count = 0

while True:
    cur_row = curs.fetchone()
    if cur_row: 
        row_count += 1
        print (cur_row)
    else:
        print("No, records.") 
        break

print("Total of Rows", end=": ") 
print (row_count)
```

```python
row_count = 0

for cur_row in curs.fetchall():
    for row_field in cur_row:
        if isinstance(row_field, float): 
           print ('%f' % row_field, end="; ")
        else:
           print (row_field, end="; ")
    row_count += 1
    print ("")

print("Total of Rows", end=": ") 
print (row_count)
```
