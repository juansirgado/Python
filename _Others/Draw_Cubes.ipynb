```python
#----------------------------------------------------------#
t = turtle.Turtle()
t.speed(0)
x = 100
c = ['Red','Green','Blue','Yellow','Orange','Pink','Purple','Brown','White','Black',]
#----------------------------------------------------------#

screen = turtle.Screen()  # Grab/create a resource e.g. a window
screen.clear()
screen.title("Draw my Cubes")
screen.screensize(680,680)

for i in range(100):
    t.left(10-(i%3))
    t.color(c[i%3])
    for j in range(4):
        t.forward(x-i)
        t.left(90)

turtle.done()
time.sleep(10)
#----------------------------------------------------------#
```

```python
c = ['red', 'purple', 'blue', 'green', 'orange', 'yellow']
t = turtle.Pen()

screen.clear()
turtle.bgcolor('black')

for x in range(360):
    t.pencolor(c[x%6])
    t.width(x//100 + 1)
    t.forward(x)
    t.left(59)

turtle.done()
time.sleep(10)    
```
