```python
import os
import folium

m = folium.Map([37, 0], zoom_start=2, tiles="stamentoner")
merc = os.path.join("ImageOverlay.png")


if not os.path.isfile(merc):
    print(f"Could not find {merc}")
else:
    img = folium.raster_layers.ImageOverlay(
        name="USGS",
        image=merc,
        bounds=[[100, -200], [80, -140]],
        opacity=1,
        interactive=True,
        cross_origin=False,
        zindex=1)

    folium.Popup("https://earthquake.usgs.gov").add_to(img)

    img.add_to(m)
    folium.LayerControl().add_to(m)

m
```
