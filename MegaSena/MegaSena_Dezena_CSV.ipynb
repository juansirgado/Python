```python
# CVS to CSVDataset MegaSena
from pandas import read_csv

def readCSV(file_in):
    df_in = read_csv(file_in, index_col="Concurso")
    df_in.sort_values(by=["Concurso"],ascending=True,inplace=True)
    return df_in
```

```python
# Convert the CSV file
import numpy as np

def convertCSV(df_in):

    array_line = ['Sequence','Date','Value1','Value2','Value3','Value4','Value5','Value6']
    work_array = np.array(array_line)

    for dfinterrows in df_in.iterrows():
        invdate = str(dfinterrows[1][0])[6:10] + "-" + str(dfinterrows[1][0])[3:5] + "-" + str(dfinterrows[1][0])[0:2]
        array_row = [dfinterrows[0], invdate,
                    dfinterrows[1][1], dfinterrows[1][2], dfinterrows[1][3], 
                    dfinterrows[1][4], dfinterrows[1][5], dfinterrows[1][6]]
        work_array = np.vstack([work_array, array_row])
    return work_array
```

```python
# Write the CSV file
import pandas as pd

def writeCSV(work_array,file_name):
    df_work = pd.DataFrame(work_array)
    file_out = file_name
    df_work.to_csv(file_out, index=False,header=False)
    return df_work.count()
```

```python
# Read/Convert/Write the CSV file
files_in = ["mega_sena_sorteio.csv","mega_sena_crescente.csv"]
file_names = ["MegaSena_Dezena_SRT.csv","MegaSena_Dezena_CRS.csv"]

for file_in in files_in:
    df_file = readCSV(file_in)
    work_array = convertCSV(df_file)
    file_size = writeCSV(work_array, file_names)
    print("Total records writted: " + str(file_size))
```
