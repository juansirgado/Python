```python
card_count = 0

for dezena_01 in range(1, 56):
    print(f'Dezena 01: {dezena_01:5.0f} Cards: {card_count:15.0f}')
    for dezena_02 in range(dezena_01 + 1, 56 + 1):
        for dezena_03 in range(dezena_02 + 1, 57 + 1):
            for dezena_04 in range(dezena_03 + 1, 58 + 1):
                for dezena_05 in range(dezena_04 + 1, 59 + 1):
                    for dezena_06 in range(dezena_05 + 1, 60 + 1):
                        card_count += 1
                        card_sum = dezena_01 + dezena_02 + dezena_03 + dezena_04 + dezena_05 + dezena_06
                        if card_sum == 30:
                            print(f'Card: {card_count:15.0f} Numbers: {dezena_01:02.0f} {dezena_02:02.0f} {dezena_03:02.0f} {dezena_04:02.0f} {dezena_05:02.0f} {dezena_06:02.0f} ')
                        if dezena_01 == 1 and dezena_02 == 2 and dezena_03 == 3 and dezena_04 == 4 and dezena_05 == 5 and dezena_06 == 6:
                            print(f'Card: {card_count:15.0f} Numbers: {dezena_01:02.0f} {dezena_02:02.0f} {dezena_03:02.0f} {dezena_04:02.0f} {dezena_05:02.0f} {dezena_06:02.0f} ')

print(f'Final total cards: {card_count:15.0f}')
```
